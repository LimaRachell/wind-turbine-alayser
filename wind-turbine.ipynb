```python
import pandas as pd
import numpy as np

# ============================================================
# HAI 21.03
# ============================================================

BASE = "/kaggle/input/datasets/icsdataset/hai-security-dataset/hai-21.03"

# Normal training files
train1 = pd.read_csv(f"{BASE}/train1.csv")
train2 = pd.read_csv(f"{BASE}/train2.csv")
train3 = pd.read_csv(f"{BASE}/train3.csv")

# Attack/test files
test1 = pd.read_csv(f"{BASE}/test1.csv")
test2 = pd.read_csv(f"{BASE}/test2.csv")
test3 = pd.read_csv(f"{BASE}/test3.csv")
test4 = pd.read_csv(f"{BASE}/test4.csv")
test5 = pd.read_csv(f"{BASE}/test5.csv")

print("FILES LOADED")
print("-------------------------")

for name, df in [
    ("train1", train1),
    ("train2", train2),
    ("train3", train3),
    ("test1", test1),
    ("test2", test2),
    ("test3", test3),
    ("test4", test4),
    ("test5", test5)
]:
    print(f"{name:7s}: {df.shape}")
```

```python
# ============================================================
# NORMAL DATA
# ============================================================

normal_data = pd.concat(
    [train1, train2, train3],
    ignore_index=True
)

# ============================================================
# ATTACK DATA
# Use test1-test4 ONLY
# ============================================================

attack_source = pd.concat(
    [test1, test2, test3, test4],
    ignore_index=True
)

attack_data = attack_source[
    attack_source["attack"] == 1
].copy()

print("Normal samples available :", len(normal_data))
print("Attack samples available :", len(attack_data))
```

```python
# ============================================================
# BALANCE NORMAL AND ATTACK
# ============================================================

n_attack = len(attack_data)

normal_sample = normal_data.sample(
    n=n_attack,
    random_state=42
)

# Combine
rf_train = pd.concat(
    [normal_sample, attack_data],
    ignore_index=True
)

# Shuffle
rf_train = rf_train.sample(
    frac=1,
    random_state=42
).reset_index(drop=True)

print("\nBalanced training dataset:")
print(rf_train["attack"].value_counts())

print("\nShape:")
print(rf_train.shape)
```

```python
# ============================================================
# LABEL COLUMNS
# ============================================================

LABEL_COLUMNS = [
    "attack",
    "attack_P1",
    "attack_P2",
    "attack_P3"
]

# ============================================================
# TRAINING FEATURES
# ============================================================

X_train = rf_train.drop(
    columns=LABEL_COLUMNS
)

y_train = rf_train["attack"]

# ============================================================
# FINAL TEST = test5 ONLY
# ============================================================

X_test = test5.drop(
    columns=LABEL_COLUMNS
)

y_test = test5["attack"]

print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("X_test :", X_test.shape)
print("y_test :", y_test.shape)
```

```python
# ============================================================
# FIND NON-NUMERIC COLUMNS
# ============================================================

object_columns = X_train.select_dtypes(
    include=["object"]
).columns.tolist()

print("Object columns:")
print(object_columns)
```

```python
for col in object_columns:
    print("\n================================")
    print("COLUMN:", col)
    print("================================")
    print(X_train[col].head())
```

```python
# ============================================================
# CONVERT TO NUMERIC
# ============================================================

X_train = X_train.apply(
    pd.to_numeric,
    errors="coerce"
)

X_test = X_test.apply(
    pd.to_numeric,
    errors="coerce"
)

print("Train dtypes:")
print(X_train.dtypes.value_counts())
```

```python
print("Number of training features:", X_train.shape[1])
print("Number of testing features :", X_test.shape[1])

print("Same columns:",
      list(X_train.columns) == list(X_test.columns))
```

```python
# ============================================================
# INFINITY → NaN
# ============================================================

X_train = X_train.replace(
    [np.inf, -np.inf],
    np.nan
)

X_test = X_test.replace(
    [np.inf, -np.inf],
    np.nan
)

# ============================================================
# MEDIAN IMPUTATION
# Calculate ONLY from training data
# ============================================================

train_medians = X_train.median()

X_train = X_train.fillna(train_medians)
X_test = X_test.fillna(train_medians)

print("Missing values in X_train:",
      X_train.isna().sum().sum())

print("Missing values in X_test:",
      X_test.isna().sum().sum())
```

```python
# ============================================================
# CHECK FEATURE MATCHING
# ============================================================

print("Train features:", X_train.shape[1])
print("Test features :", X_test.shape[1])

print(
    "Same features:",
    list(X_train.columns) == list(X_test.columns)
)
```

```python
print("TRAIN LABELS")
print(y_train.value_counts())

print("\nFINAL TEST LABELS")
print(y_test.value_counts())
```

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    n_jobs=-1,
    class_weight="balanced"
)

print("Training Random Forest...")

rf.fit(
    X_train,
    y_train
)

print("Training completed!")
```
